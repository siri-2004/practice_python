## Datatype conversions

```python
# float to int
a = 7
b = 5.7
print(a)
print(b)
c = a+b
print(c)
print(type(c))
c = int(c)
print(c)
print(type(c))
```

```python
# int to float
a = -57
print(a)
print(type(a))
a = float(a)
print(a)
print(type(a))
```

```python
# str to int
a = '56'
print(a)
print(type(a))
a = int(a)
print(a)
print(type(a))
```

```python
# str to list
a = '43'
print(a)
print(type(a))
a = list(a)
print(a)
print(type(a))
```

```python
# str to tuple
a = '43'
print(a)
print(type(a))
a = tuple(a)
print(a)
print(type(a))
```

```python
# str to set
a = '434563'
print(a)
print(type(a))
a = set(a)
print(a)
print(type(a))
print(type(a).__name__) # it does not give <class 'set'>, it just gives set (datatype name)
```

```python
# str to dict ---> not possible, because it takes key value pairs
a = '43'
print(a)
print(type(a))
a = dict(a)
print(a)
print(type(a))
```

```python
# str to float ---> not possible because, it has str value
a = 'abc'
print(a)
print(type(a))
a = float(a)
print(a)
print(type(a))
rint(type(a).__name__
```
