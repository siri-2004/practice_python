```python
#arithmetic operators +, - , /, //, %, *

a=10
b=3
c = a+b
print(c)
d = a-b
print(d)
e = a/b #division
print(e)
f = a//b #floor division
print(f)
g = a%b #modulus- reminder
print(g)
h = a*b #multiply
print(h)
i = a**b #exponent/power
print(i)
```

```python
#assignment operators =, +=, -=, /=, *=

a = 3
a +=2
print(a)
b = 6
b -= 2
print(b)
c = 8
c /= 2
print(c)
d = 7
d //= 2
print(d)
e = 7
e %= 4
print(e)
f = 5
f *= 2
print(f)
g = 6
g **= 2
print(g)
```

```python
#comparison operators ==, !=, >, <, >=, <=
a,b = 3,5
print(a==b)
print(a!=b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
```

```python
#logical operators and, or, not

a,b = True, False
print(a and b)
print(a or b)
print(not a)
print(not b)
```

```python
#identity operators is, is not

list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a

print(list_a is list_b)
print(list_a is not list_b)
print(list_a is list_c)
print(list_a is not list_c)
```
